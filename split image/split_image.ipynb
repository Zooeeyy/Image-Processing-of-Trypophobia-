```python
import os
import shutil
from sklearn.model_selection import train_test_split

#train directory
data_dir = 'C:/Data Mining/skrips/trypophobia/trypophobia/train/norm'

# new directory path for saving data train and validatiom
train_dir = 'C:/Data Mining/skrips/trypophobia/tryp/trypo_train/norm'
validation_dir = 'C:/Data Mining/skrips/trypophobia/tryp/trypo_valid/norm'

image_files = [os.path.join(data_dir, f) for f in os.listdir(data_dir) if f.endswith('.png')]

#Split data with ratio 80:20
train_files, validation_files = train_test_split(image_files, test_size=0.2, train_size=0.8, random_state=42)

#save 
for file in train_files:
    shutil.copy(file, os.path.join(train_dir, os.path.basename(file)))

for file in validation_files:
    shutil.copy(file, os.path.join(validation_dir, os.path.basename(file)))
```
